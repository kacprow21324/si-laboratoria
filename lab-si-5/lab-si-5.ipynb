```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
plt.style.use('fivethirtyeight')
import pandas as pd
from keras.callbacks import EarlyStopping
from sklearn.preprocessing import MinMaxScaler
from keras.models import Sequential
from keras.layers import Dense, LSTM, Dropout, GRU, Bidirectional, Input
from keras.optimizers import Adam, Nadam
AdamW = tf.keras.optimizers.AdamW
import math
from sklearn.metrics import mean_squared_error, mean_absolute_error, mean_absolute_percentage_error
from keras.losses import Huber
MSLE = tf.keras.losses.MeanSquaredLogarithmicError
```

```python
def plot_predictions(test, predicted, name='GOOGL'):
    plt.figure(figsize=(16, 8))
    plt.plot(test, color='red', label='Rzeczywista')
    plt.plot(predicted, color='blue', label='Przewidywana')
    plt.title(name)
    plt.xlabel('Czas')
    plt.ylabel('Cena [USD]')
    plt.legend()
    plt.show()

def return_rmse(test, predicted):
    rmse = math.sqrt(mean_squared_error(test, predicted))
    print('-' * 30)
    print('Blad sredniokwadratowy modelu {:.2f}.'.format(rmse))
    print('-' * 30)

def return_mae(test, predicted):
    mae = mean_absolute_error(test, predicted)
    mape = mean_absolute_percentage_error(test, predicted) * 100
    print('Sredni blad bezwzgledny: {:.2f}.'.format(mae))
    print('Sredni blad bezwzgledny procentowy: {:.2f}%'.format(mape))
    print('-' * 30)
```

```python
dataset = pd.read_csv('GOOGL_2006-01-01_to_2018-01-01.csv', index_col='Date', parse_dates=['Date'])
dataset.head()
```

```python
# Ustawiamy zbiory trenujące i testowe oraz szukamy brakujących wartości
training_set = dataset[:'2016'].iloc[:,1:2].values
test_set = dataset['2017':].iloc[:,1:2].values
```

```python
# Wizualizacja danych
dataset["High"][:'2016'].plot(figsize=(16, 8),legend=True)
dataset["High"]['2017':].plot(figsize=(16, 8),legend=True)
plt.legend(['Trening (do 2016)', 'Test (2017+)'])
plt.title('GOOGL')
plt.show()
```

```python
# Normalizacja danych trenujących
sc = MinMaxScaler(feature_range=(0,1))
training_set_scaled = sc.fit_transform(training_set)
```

```python
X_train = []
y_train = []
for i in range(60, len(training_set_scaled)):
    X_train.append(training_set_scaled[i-60:i, 0])
    y_train.append(training_set_scaled[i, 0])
X_train, y_train = np.array(X_train), np.array(y_train)
```

```python
# Przekształcanie X_train w celu zapewnienia wydajnego modelowania
X_train = np.reshape(X_train, (X_train.shape[0],X_train.shape[1],1))
```

```python
regressor = Sequential()
regressor.add(Input(shape=(X_train.shape[1], 1)))
regressor.add(LSTM(units=100, return_sequences=True))
regressor.add(Dropout(0.2))
regressor.add(LSTM(units=100, return_sequences=True))
regressor.add(Dropout(0.2))
regressor.add(LSTM(units=100, return_sequences=True))
regressor.add(Dropout(0.2))
regressor.add(LSTM(units=100))
regressor.add(Dropout(0.2))
regressor.add(Dense(units=1))
regressor.compile(optimizer='rmsprop', loss='mean_squared_error')
regressor.fit(X_train, y_train, epochs=50, batch_size=32)
```

```python
dataset_total = pd.concat((dataset['High'][:'2016'], dataset['High']['2017':]), axis=0)
inputs = dataset_total[len(dataset_total)-len(test_set) - 60:].values
inputs = inputs.reshape(-1, 1)
inputs = sc.transform(inputs)
```

```python
X_test = []
for i in range(60, len(inputs)):
    X_test.append(inputs[i-60:i, 0])
X_test = np.array(X_test)
X_test = np.reshape(X_test, (X_test.shape[0], X_test.shape[1], 1))
predicted_stock_price = regressor.predict(X_test)
predicted_stock_price = sc.inverse_transform(predicted_stock_price)
```

```python
plot_predictions(test_set, predicted_stock_price, 'LSTM 500 jednostek, rmsprop, MSE')
```

```python
return_rmse(test_set, predicted_stock_price)
return_mae(test_set, predicted_stock_price)
```

```python
regressorGRU = Sequential()
regressorGRU.add(Input(shape=(X_train.shape[1], 1)))
regressorGRU.add(GRU(units=50, return_sequences=True, activation='tanh'))
regressorGRU.add(Dropout(0.2))
regressorGRU.add(GRU(units=50, return_sequences=True, activation='tanh'))
regressorGRU.add(Dropout(0.2))
regressorGRU.add(GRU(units=50, return_sequences=True, activation='tanh'))
regressorGRU.add(Dropout(0.2))
regressorGRU.add(GRU(units=50, activation='tanh'))
regressorGRU.add(Dropout(0.2))
regressorGRU.add(Dense(units=1))
regressorGRU.compile(optimizer='rmsprop', loss='mean_squared_error')
early_stop = EarlyStopping(monitor='loss', patience=5, restore_best_weights=True)
regressorGRU.fit(X_train, y_train, epochs=100, batch_size=32, callbacks=[early_stop])
```

```python
X_test = []
for i in range(60, len(inputs)):
    X_test.append(inputs[i-60:i, 0])
X_test = np.array(X_test)
X_test = np.reshape(X_test, (X_test.shape[0], X_test.shape[1], 1))
GRU_predicted_stock_price = regressorGRU.predict(X_test)
GRU_predicted_stock_price = sc.inverse_transform(GRU_predicted_stock_price)
```

```python
plot_predictions(test_set, GRU_predicted_stock_price, 'GRU 50 jednostek, rmsprop, MSE, Early Stopping')
```

```python
return_rmse(test_set, GRU_predicted_stock_price)
return_mae(test_set, GRU_predicted_stock_price)
```

```python
def run_experiment(units=100, optimizer='rmsprop', loss_fn='mean_squared_error',
                   epochs=50, validation_split=0.0, early_stop_monitor=None, label='',
                   X_tr=None, y_tr=None, sc_used=None, test_s=None, inputs_used=None,
                   patience=5, dropout=0.2):
    X_tr = X_tr if X_tr is not None else X_train
    y_tr = y_tr if y_tr is not None else y_train
    sc_used = sc_used if sc_used is not None else sc
    test_s = test_s if test_s is not None else test_set
    inputs_used = inputs_used if inputs_used is not None else inputs
    model = Sequential([
        Input(shape=(X_tr.shape[1], 1)),
        LSTM(units=units, return_sequences=True),
        Dropout(dropout),
        LSTM(units=units, return_sequences=True),
        Dropout(dropout),
        LSTM(units=units, return_sequences=True),
        Dropout(dropout),
        LSTM(units=units),
        Dropout(dropout),
        Dense(units=1)
    ])
    model.compile(optimizer=optimizer, loss=loss_fn)
    cbs = [EarlyStopping(monitor=early_stop_monitor, patience=patience, restore_best_weights=True)] if early_stop_monitor else []
    history = model.fit(X_tr, y_tr, epochs=epochs, batch_size=32,
                        validation_split=validation_split,
                        callbacks=cbs if cbs else None, verbose=1)
    X_t = []
    for i in range(60, len(inputs_used)):
        X_t.append(inputs_used[i-60:i, 0])
    X_t = np.array(X_t).reshape(-1, X_tr.shape[1], 1)
    pred = sc_used.inverse_transform(model.predict(X_t, verbose=0))
    print('=== ' + label + ' ===')
    if early_stop_monitor:
        print('Zatrzymano po epoce: ' + str(len(history.history['loss'])))
    plot_predictions(test_s, pred, label)
    return_rmse(test_s, pred)
    return_mae(test_s, pred)
    return pred, math.sqrt(mean_squared_error(test_s, pred))
```

## Zadanie A – Eksperymenty z liczbą jednostek LSTM

Porównanie modeli z 16, 32, 100, 200 i 500 jednostkami LSTM. Pozostałe parametry stałe: rmsprop, MSE, 50 epok.

```python
for units in [16, 32, 100, 200, 500]:
    run_experiment(units=units, label='LSTM ' + str(units) + ' jednostek')
```

## Zadanie B – Różne optymizatory

Porównanie rmsprop, Adam, AdamW i Nadam. Stałe: 50 jednostek, MSE, 50 epok.

```python
configs_B = [
    ('rmsprop', 'rmsprop'),
    ('adam', 'Adam'),
    (AdamW(), 'AdamW'),
    (Nadam(), 'Nadam'),
]
for opt, name in configs_B:
    run_experiment(units=50, optimizer=opt, label='Optimizer: ' + name)
```

## Zadanie C – Różne funkcje straty

Porównanie MSE, MAE, Huber i MSLE (Mean Squared Log Error). Stałe: 50 jednostek, Adam, 50 epok.

```python
losses_C = [
    ('mean_squared_error', 'MSE'),
    ('mean_absolute_error', 'MAE'),
    (Huber(), 'Huber'),
    (MSLE(), 'MSLE'),
]
for loss_fn, name in losses_C:
    run_experiment(units=50, optimizer='adam', loss_fn=loss_fn, label='Loss: ' + name)
```

## Zadanie D – Zmiana atrybutu predykcji: High → Close

Przebudowanie danych wejściowych z kolumną Close.

```python
training_set_D = dataset[:'2016'].iloc[:, 3:4].values
test_set_D = dataset['2017':].iloc[:, 3:4].values
sc_D = MinMaxScaler(feature_range=(0, 1))
training_set_D_scaled = sc_D.fit_transform(training_set_D)
X_train_D, y_train_D = [], []
for i in range(60, len(training_set_D_scaled)):
    X_train_D.append(training_set_D_scaled[i-60:i, 0])
    y_train_D.append(training_set_D_scaled[i, 0])
X_train_D = np.array(X_train_D).reshape(-1, 60, 1)
y_train_D = np.array(y_train_D)
dataset_total_D = pd.concat((dataset['Close'][:'2016'], dataset['Close']['2017':]), axis=0)
inputs_D = dataset_total_D[len(dataset_total_D) - len(test_set_D) - 60:].values.reshape(-1, 1)
inputs_D = sc_D.transform(inputs_D)
run_experiment(units=50, optimizer='adam', label='LSTM atrybut Close',
               X_tr=X_train_D, y_tr=y_train_D, sc_used=sc_D,
               test_s=test_set_D, inputs_used=inputs_D)
```

## Zadanie E – Early Stopping z val_loss

Porównanie 100 epok bez Early Stopping vs z Early Stopping monitorującym val_loss.

```python
run_experiment(units=50, epochs=100,
               label='LSTM bez Early Stopping (100 epok)')

run_experiment(units=50, epochs=100, validation_split=0.1,
               early_stop_monitor='val_loss',
               label='LSTM z Early Stopping val_loss')
```

## Zadanie F – RMSE poniżej 2.0

Konfiguracja: 256 jednostek + Adam(lr=5e-4) + Huber + Early Stopping val_loss (patience=10) + Dropout 0.1.

```python
run_experiment(units=256, optimizer=Adam(learning_rate=0.0005), loss_fn=Huber(),
               epochs=150, validation_split=0.1, early_stop_monitor='val_loss',
               patience=10, dropout=0.1,
               label='256j + Adam(lr=5e-4) + Huber + ES(patience=10) + Dropout 0.1')
```
